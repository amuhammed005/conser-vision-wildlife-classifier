# Conser-vision: Wildlife Camera-Trap Image Classification

**Objective.** Identify the animal, or an empty frame, in camera-trap photographs. Every image belongs to one of eight classes: `antelope_duiker`, `bird`, `blank`, `civet_genet`, `hog`, `leopard`, `monkey_prosimian` and `rodent`. Submissions are scored with **multiclass log loss** (lower is better), so well-calibrated probabilities matter as much as accuracy.

**Approach.** An ensemble of classifiers on frozen pretrained backbones plus a partially fine-tuned DINOv2, calibrated and blended:

| Stage | Description |
|---|---|
| Feature extraction | Embeddings from pretrained backbones (DINOv2, CLIP, ConvNeXt) at several input resolutions |
| Linear and MLP heads | Logistic regression and a one-hidden-layer MLP trained on the embeddings |
| Fine-tuning | Last four DINOv2 ViT-B blocks trained with strong augmentation, one model per resolution |
| Calibration and blending | Per-model temperature scaling, blend weights searched on out-of-fold predictions, final calibration |

**Validation.** Five-fold `StratifiedGroupKFold`, grouped by camera site. The test images come from cameras that never appear in the training data, so keeping each site inside a single fold makes the out-of-fold (OOF) score an honest estimate of performance on unseen cameras.

**Requirements.** A GPU and internet access (pretrained weights are downloaded on first use). Every stage writes a checkpoint and is skipped when its file already exists, so an interrupted run resumes where it stopped.

```python
# Image resolution and greyscale (infrared night) share, measured on a random sample
def get_image_info(path):
    """Read an image's size and test whether it is (nearly) greyscale.

    Greyscale frames are usually infrared night captures.

    Args:
        path: Path to a JPEG image.

    Returns:
        Tuple ``(width, height, is_grey)``.
    """
    img = Image.open(path)
    width, height = img.size            # original size, read before the reduced-size decode
    img.draft("RGB", (160, 90))         # fast reduced-size JPEG decode
    small = np.asarray(img.convert("RGB"), dtype=np.int16)
    # channels are nearly equal in greyscale frames
    colour_gap = np.abs(small[:, :, 0] - small[:, :, 1]).mean() + np.abs(small[:, :, 1] - small[:, :, 2]).mean()
    return width, height, bool(colour_gap < 3)    # heuristic threshold


sample_idx = np.random.default_rng(SEED).choice(len(train_paths), size=min(1500, len(train_paths)), replace=False)

rows = []
for i in tqdm(sample_idx, desc="Checking sample images"):
    width, height, is_grey = get_image_info(train_paths[i])
    rows.append({"width": width, "height": height, "is_grey": is_grey, "class": species_labels[y[i]]})
image_info = pd.DataFrame(rows)

image_info["size"] = image_info["width"].astype(str) + "x" + image_info["height"].astype(str)
print("Image sizes in the sample:")
print(image_info["size"].value_counts().head(10))

print("\nShare of greyscale (likely night) images:", round(image_info["is_grey"].mean(), 2))
grey_by_class = image_info.groupby("class")["is_grey"].mean().sort_values()
print(grey_by_class.round(2))

grey_by_class.plot(kind="barh", figsize=(8, 4))
plt.title("Share of greyscale images per class (sample)")
plt.xlabel("Share greyscale")
plt.show()
```

## Step 4: Feature Extraction
A pretrained backbone turns each photograph into a vector of numbers, its *embedding*. Embeddings are extracted for several backbones and input sizes, each averaged with its horizontally flipped view, and cached to disk so nothing is computed twice. A backbone that fails (invalid name or insufficient GPU memory) is skipped and the run continues.

```python
class ImageDataset(Dataset):
    """Map-style dataset that loads an image from disk and applies a transform.

    Each item is an ``(image_tensor, label)`` tuple.

    Args:
        paths: Image file paths.
        labels: Integer class ids aligned with ``paths`` (zeros for unlabelled test images).
        transform: Callable applied to each PIL image.
    """

    def __init__(self, paths, labels, transform):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, i):
        img = Image.open(self.paths[i]).convert("RGB")
        return self.transform(img), int(self.labels[i])


@torch.no_grad()
def extract_features(model, loader, desc="Extracting"):
    """Embed every image, averaging the original and horizontally flipped views.

    Args:
        model: Backbone returning pooled embeddings (no classification head).
        loader: DataLoader yielding ``(images, labels)`` batches in a fixed order.
        desc: Progress-bar label.

    Returns:
        ``np.ndarray`` of shape ``(n_images, embedding_dim)``.
    """
    model.eval()
    all_feats = []
    for batch, _ in tqdm(loader, desc=desc, leave=False):
        batch = batch.to(device)
        with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
            feats_normal = model(batch)
            feats_flipped = model(torch.flip(batch, dims=[3]))   # horizontal flip
        all_feats.append(((feats_normal + feats_flipped) / 2).float().cpu())
    return torch.cat(all_feats).numpy()
```

```python
def get_features(backbone_name, img_size):
    """Return train and test embeddings for one backbone at one input size.

    Features are cached to ``features_<backbone>_<size>.npz`` and reloaded on later runs.

    Args:
        backbone_name: timm model identifier.
        img_size: Square input resolution in pixels.

    Returns:
        Tuple ``(train_features, test_features)``.
    """
    safe_name = backbone_name.replace(".", "_").replace("/", "_")
    cache_file = f"features_{safe_name}_{img_size}.npz"

    # Reuse cached features when available
    if os.path.exists(cache_file):
        print("Loaded from cache:", cache_file)
        saved = np.load(cache_file)
        return saved["train"], saved["test"]

    # ViTs need img_size set explicitly; convnets accept any input size
    extra = {"img_size": img_size} if "vit" in backbone_name else {}
    model = timm.create_model(backbone_name, pretrained=True, num_classes=0, **extra)
    model.to(device)

    # Resize to a square and normalise with the backbone's own statistics
    cfg = timm.data.resolve_data_config({}, model=model)
    tfm = transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=cfg["mean"], std=cfg["std"]),
    ])

    train_ds = ImageDataset(train_paths, y, tfm)
    test_ds = ImageDataset(test_paths, np.zeros(len(test_paths), dtype=int), tfm)
    train_loader = DataLoader(train_ds, batch_size=EXTRACT_BATCH_SIZE, shuffle=False,
                              num_workers=NUM_WORKERS, pin_memory=True)
    test_loader = DataLoader(test_ds, batch_size=EXTRACT_BATCH_SIZE, shuffle=False,
                             num_workers=NUM_WORKERS, pin_memory=True)

    start = time.time()
    train_f = extract_features(model, train_loader, desc=f"{backbone_name}@{img_size} train")
    test_f = extract_features(model, test_loader, desc=f"{backbone_name}@{img_size} test")
    # Cache to disk
    np.savez(cache_file, train=train_f, test=test_f)
    print(f"Saved {cache_file} ({(time.time() - start) / 60:.1f} min)")

    # Release GPU memory before the next backbone loads
    del model
    gc.collect()
    torch.cuda.empty_cache()
    return train_f, test_f
```

```python
# Extract (or load from cache) features for every backbone / size pair
feature_sets = {}
for name, size in tqdm(FEATURE_SPECS, desc="Feature sets"):
    key = f"{name}@{size}"
    try:
        feature_sets[key] = get_features(name, size)
        print(key, "-> train", feature_sets[key][0].shape, "| test", feature_sets[key][1].shape)
    except Exception as e:
        # A failing backbone (invalid name, out of memory) is skipped rather than aborting the run
        print("!! skipped", key, "because of:", repr(e))
        gc.collect()
        torch.cuda.empty_cache()

print("\nFeature sets ready:", len(feature_sets), "of", len(FEATURE_SPECS))
```

## Step 5: Logistic Regression
Logistic regression on standardised features (`make_pipeline(StandardScaler(), LogisticRegression())`) is the first classifier. It is selected in three stages:

1. Score every feature set on its own.
2. Test whether stacking the best sets beats the best single set.
3. Tune the regularisation strength `C` on the winner.

```python
def score(probs):
    """Multiclass log loss of predictions against the training labels.

    Rows are renormalised first to remove floating-point drift.

    Args:
        probs: Array of shape ``(n_train, n_classes)``.

    Returns:
        Log loss (lower is better).
    """
    probs = np.asarray(probs, dtype=np.float64)
    probs = probs / probs.sum(axis=1, keepdims=True)    # removes tiny rounding errors that trigger warnings
    return log_loss(y, probs, labels=list(range(N_CLASSES)))


def stack(keys, part):
    """Concatenate the embeddings of several feature sets column-wise.

    Args:
        keys: Keys of ``feature_sets`` to combine.
        part: ``0`` for training features, ``1`` for test features.

    Returns:
        Array of shape ``(n_images, total_dim)``.
    """
    return np.hstack([feature_sets[k][part] for k in keys])


def run_cv_lr(X, C):
    """Out-of-fold predictions from a standardised logistic regression.

    Trains on four folds and predicts the fifth, rotating through all folds.

    Args:
        X: Training feature matrix.
        C: Inverse regularisation strength.

    Returns:
        Array of shape ``(n_train, n_classes)`` with out-of-fold probabilities.
    """
    oof = np.zeros((len(y), N_CLASSES))
    fold_bar = tqdm(folds, desc=f"LogReg C={C}", leave=False)
    for fold_num, (tr_idx, va_idx) in enumerate(fold_bar, start=1):
        model = make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=1000))
        model.fit(X[tr_idx], y[tr_idx])
        oof[va_idx] = model.predict_proba(X[va_idx])

        # show each fold's score as soon as it finishes
        fold_ll = log_loss(y[va_idx], oof[va_idx], labels=list(range(N_CLASSES)))
        fold_bar.set_postfix(fold=fold_num, log_loss=f"{fold_ll:.3f}")
    return oof
```

```python
# Checkpoint files for this stage
LR_FILE = "part3_lr.npz"
LR_SETTINGS_FILE = "part3_lr_settings.json"
LR_C_START = 0.004      # regularisation strength used to rank the feature sets


def short_name(key):
    """Drop the pretrained-weights tag from a feature-set key.

    Example: ``'vit_base_patch14_dinov2.lvd142m@448'`` becomes ``'vit_base_patch14_dinov2@448'``.

    Args:
        key: Feature-set key of the form ``'<model>@<size>'``.

    Returns:
        Shortened key.
    """
    model_part, size = key.split("@")
    return model_part.split(".")[0] + "@" + size


# Resume from the checkpoint when the search has already run
if os.path.exists(LR_FILE) and os.path.exists(LR_SETTINGS_FILE):
    print("Loading saved logistic regression results")
    with open(LR_SETTINGS_FILE) as f:
        lr_settings = json.load(f)
    saved = np.load(LR_FILE)
    oof_lr, test_lr = saved["oof"], saved["test"]

else:
    # (a) Score each feature set on its own
    single_scores = {}
    for key in tqdm(list(feature_sets), desc="Single feature sets"):
        oof = run_cv_lr(feature_sets[key][0], C=LR_C_START)
        single_scores[key] = score(oof)
        tqdm.write(f"{key:55s} OOF log loss {single_scores[key]:.4f}")
    ranked = sorted(single_scores, key=single_scores.get)

    # (b) Compare stacking the top-k feature sets with the best single set (same C)
    candidates = {"best single": ranked[:1]}
    for k in [2, 3, 4, len(ranked)]:
        if 2 <= k <= len(ranked):
            candidates[f"top {k} combined"] = ranked[:k]

    cand_scores = {"best single": single_scores[ranked[0]]}
    for cand_name, keys in tqdm(candidates.items(), desc="Combinations"):
        if cand_name == "best single":
            continue
        cand_scores[cand_name] = score(run_cv_lr(stack(keys, 0), C=LR_C_START))
        tqdm.write(f"{cand_name:20s} OOF log loss {cand_scores[cand_name]:.4f}")

    best_cand = min(cand_scores, key=cand_scores.get)
    best_keys = candidates[best_cand]
    print("\nBest choice:", best_cand, "->", best_keys)

    # (c) Tune the regularisation strength C
    X_best = stack(best_keys, 0)
    C_oofs = {}
    for C in tqdm([0.002, 0.004, 0.007], desc="Tuning C"):
        C_oofs[C] = run_cv_lr(X_best, C)
        tqdm.write(f"C = {C:<7} OOF log loss {score(C_oofs[C]):.4f}")
    best_C = min(C_oofs, key=lambda c: score(C_oofs[c]))
    oof_lr = C_oofs[best_C]
    print("Best C:", best_C, "| OOF log loss:", round(score(oof_lr), 4))

    # (d) Refit on all training data and predict the test set
    print("Fitting the final logistic regression on all training images...")
    start = time.time()
    final_lr = make_pipeline(StandardScaler(), LogisticRegression(C=best_C, max_iter=1000))
    final_lr.fit(X_best, y)
    test_lr = final_lr.predict_proba(stack(best_keys, 1))
    print(f"done in {time.time() - start:.0f}s")

    lr_settings = {"best_keys": best_keys, "best_C": best_C, "ranked": ranked}
    np.savez(LR_FILE, oof=oof_lr, test=test_lr)
    with open(LR_SETTINGS_FILE, "w") as f:
        json.dump(lr_settings, f, indent=2)
    print("Saved", LR_FILE, "and", LR_SETTINGS_FILE)

best_keys = lr_settings["best_keys"]
best_C = lr_settings["best_C"]
ranked = lr_settings["ranked"]
X_all = stack(best_keys, 0)
X_test_all = stack(best_keys, 1)
print("Features used:", best_keys)
print("Logistic regression OOF log loss (raw):", round(score(oof_lr), 4))
```

```python
def fit_lr_component(keys, C):
    """Fit logistic regression on one feature set (or a stack) as a blend component.

    Out-of-fold and test probabilities are cached to disk.

    Args:
        keys: Feature-set keys to stack.
        C: Inverse regularisation strength.

    Returns:
        Tuple ``(oof_probs, test_probs)``.
    """
    cache = "part3_lrx_" + "_".join(short_name(k).replace("@", "_") for k in keys) + ".npz"
    if os.path.exists(cache):
        saved = np.load(cache)
        return saved["oof"], saved["test"]

    oof = run_cv_lr(stack(keys, 0), C)
    model = make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=1000))
    model.fit(stack(keys, 0), y)
    test = model.predict_proba(stack(keys, 1))
    np.savez(cache, oof=oof, test=test)
    return oof, test


# The best stack above is one view of the images. The next-best feature sets are kept as
# separate blend components because models that make different mistakes blend better.
lr_components = {"logreg": (oof_lr, test_lr)}

# Three next-best feature sets, fitted with the tuned C
extra_keys = [k for k in ranked if [k] != best_keys][:3]
for key in tqdm(extra_keys, desc="Extra logistic regressions"):
    oof_x, test_x = fit_lr_component([key], best_C)
    lr_components["logreg " + short_name(key)] = (oof_x, test_x)
    tqdm.write(f"{short_name(key):45s} OOF log loss (raw) {score(oof_x):.4f}")

print("\nLogistic regression ingredients:", list(lr_components))
```

## Step 6: MLP Classifier
A one-hidden-layer MLP (512 units) on the best feature set can capture patterns a linear model cannot. `alpha` is its L2 regularisation strength; larger values give a simpler model. MLPs tend to be overconfident, so their raw log loss can look poor until calibrated in Step 7.

```python
def run_cv_mlp(X, alpha):
    """Out-of-fold predictions from a standardised one-hidden-layer MLP.

    Args:
        X: Training feature matrix.
        alpha: L2 regularisation strength.

    Returns:
        Array of shape ``(n_train, n_classes)`` with out-of-fold probabilities.
    """
    oof = np.zeros((len(y), N_CLASSES))
    fold_bar = tqdm(folds, desc=f"MLP alpha={alpha}", leave=False)
    for fold_num, (tr_idx, va_idx) in enumerate(fold_bar, start=1):
        model = make_pipeline(
            StandardScaler(),
            MLPClassifier(hidden_layer_sizes=(512,), alpha=alpha, batch_size=256,
                          max_iter=40, random_state=SEED),
        )
        model.fit(X[tr_idx], y[tr_idx])
        proba = model.predict_proba(X[va_idx]).astype(np.float64)
        oof[va_idx] = proba / proba.sum(axis=1, keepdims=True)    # renormalise: float32 rounding leaves rows not summing exactly to 1

        fold_ll = log_loss(y[va_idx], oof[va_idx], labels=list(range(N_CLASSES)))
        fold_bar.set_postfix(fold=fold_num, log_loss=f"{fold_ll:.3f}")
    return oof
```

```python
# Resume from the checkpoint when this stage has already run
MLP_FILE = "part3_mlp.npz"

if os.path.exists(MLP_FILE):
    print("Loading saved MLP results")
    saved = np.load(MLP_FILE)
    oof_mlp, test_mlp = saved["oof"], saved["test"]

else:
    alpha_oofs = {}
    for alpha in tqdm([10.0, 30.0, 100.0], desc="Tuning alpha"):   # regularisation sweep
        alpha_oofs[alpha] = run_cv_mlp(X_all, alpha)
        tqdm.write(f"alpha = {alpha:<6} OOF log loss (raw) {score(alpha_oofs[alpha]):.4f}")
    best_alpha = min(alpha_oofs, key=lambda a: score(alpha_oofs[a]))
    oof_mlp = alpha_oofs[best_alpha]
    print("Best alpha:", best_alpha)

    # Refit on all training data with the best alpha
    print("Fitting the final MLP on all training images...")
    start = time.time()
    final_mlp = make_pipeline(
        StandardScaler(),
        MLPClassifier(hidden_layer_sizes=(512,), alpha=best_alpha, batch_size=256,
                      max_iter=40, random_state=SEED),
    )
    final_mlp.fit(X_all, y)
    test_mlp = final_mlp.predict_proba(X_test_all).astype(np.float64)
    test_mlp = test_mlp / test_mlp.sum(axis=1, keepdims=True)
    print(f"done in {time.time() - start:.0f}s")

    np.savez(MLP_FILE, oof=oof_mlp, test=test_mlp)
    print("Saved", MLP_FILE)

print("MLP OOF log loss (raw):", round(score(oof_mlp), 4))
```

## Step 7: Probability Calibration
Log loss punishes confident wrong answers, so each model's probabilities receive two corrections fitted on its out-of-fold predictions:

- **Temperature** softens (above 1) or sharpens (below 1) the predictions.
- **Epsilon** mixes in a small uniform component so that no class gets a probability near zero.

```python
def calibrate(probs, temp, eps):
    """Apply temperature scaling, then mix in a small uniform component.

    Args:
        probs: Class probabilities of shape ``(n, n_classes)``.
        temp: Temperature; above 1 softens predictions, below 1 sharpens them.
        eps: Weight of the uniform distribution mixed into the result.

    Returns:
        Calibrated probabilities whose rows sum to 1.
    """
    log_p = np.log(np.clip(probs, 1e-12, 1.0))
    scaled = log_p / temp
    scaled = scaled - scaled.max(axis=1, keepdims=True)    # numerical stability for exp()
    p = np.exp(scaled)
    p = p / p.sum(axis=1, keepdims=True)
    return (1 - eps) * p + eps / N_CLASSES


def fit_calibration(probs, label="Calibration search"):
    """Grid-search the temperature and epsilon that minimise out-of-fold log loss.

    Args:
        probs: Out-of-fold probabilities.
        label: Progress-bar label.

    Returns:
        Tuple ``(temp, eps, log_loss)``.
    """
    best_temp, best_eps, best_ll = 1.0, 0.0, score(probs)
    for temp in tqdm(np.arange(0.5, 6.01, 0.25), desc=label, leave=False):
        for eps in [0.0, 0.001, 0.005, 0.01, 0.02, 0.05]:
            ll = score(calibrate(probs, temp, eps))
            if ll < best_ll:
                best_temp, best_eps, best_ll = temp, eps, ll
    if best_temp in (0.5, 6.0):
        print("   note: temp is at the edge of the search range, the range may need widening")
    return best_temp, best_eps, best_ll


# Registry of calibrated models: name -> (calibrated OOF probs, calibrated test probs)
components = {}

def add_component(name, oof, test):
    """Calibrate one model's predictions and register it for blending.

    The calibrated out-of-fold and test probabilities are stored in ``components[name]``.

    Args:
        name: Component name.
        oof: Raw out-of-fold probabilities.
        test: Raw test probabilities.
    """
    temp, eps, ll = fit_calibration(oof, label=f"Calibrating {name}")
    components[name] = (calibrate(oof, temp, eps), calibrate(test, temp, eps))
    print(f"{name:40s} raw {score(oof):.4f} -> calibrated {ll:.4f}  (temp={temp:.2f}, eps={eps})")
```

```python
# Calibrate every model trained so far
for name, (oof_c, test_c) in lr_components.items():
    add_component(name, oof_c, test_c)
add_component("mlp", oof_mlp, test_mlp)
```

## Step 8: Fine-Tuning DINOv2
The earlier stages used frozen backbones. Here the last transformer blocks are trained so the model can learn to focus on the animal rather than the camera background.

- **Frozen:** the patch embedding and all but the last `UNFREEZE_BLOCKS` blocks.
- **Trainable:** the last blocks, the final norm layer and a new 8-class head. The head learns fast and the pretrained blocks learn slowly.
- **Augmentation:** random crops that keep the photo's wide aspect ratio, horizontal flips, colour jitter, occasional greyscale (night photos are greyscale) and random erasing.
- **Sizes:** one model is fine-tuned per size in `FT_SIZES` (224 and 336 by default), and each is a separate blend component.
- **Validation:** the same site-grouped folds. Each fold saves its best weights and predictions, and the test predictions of the five fold models are averaged.
- **Early stopping:** a fold stops when its validation log loss has not improved for `FT_PATIENCE` epochs. The best epoch is chosen on the validation data itself, so the OOF score can be very slightly optimistic.
- **Loss:** cross-entropy with label smoothing (0.1), so the raw log loss is not comparable with the other models until all are calibrated.

```python
# ImageNet statistics expected by the DINOv2 weights
MEAN = timm.data.IMAGENET_DEFAULT_MEAN
STD = timm.data.IMAGENET_DEFAULT_STD


def setup_finetune(img_size, epochs):
    """Configure the fine-tuning globals for one input size.

    Sets the image size, epoch budget, checkpoint prefix and the train and eval
    transforms. Checkpoint names include the size, so each size keeps its own files.

    Args:
        img_size: Square input resolution in pixels.
        epochs: Maximum number of epochs per fold.
    """
    global FT_IMG_SIZE, FT_EPOCHS, FT_TAG, ft_train_tfm, ft_eval_tfm, ft_test_ds
    FT_IMG_SIZE = img_size
    FT_EPOCHS = epochs
    FT_TAG = f"ft_{FT_MODEL.split('.')[0]}_{img_size}_last{UNFREEZE_BLOCKS}"

    # Training augmentation prevents memorising camera backgrounds
    # (the crop ratio preserves the wide aspect of the original photos)
    ft_train_tfm = transforms.Compose([
        transforms.RandomResizedCrop(img_size, scale=(0.5, 1.0), ratio=(1.3, 2.0)),
        transforms.RandomHorizontalFlip(),
        transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.03),
        transforms.RandomGrayscale(p=0.1),
        transforms.ToTensor(),
        transforms.Normalize(MEAN, STD),
        transforms.RandomErasing(p=0.25, scale=(0.02, 0.15)),
    ])

    # Validation and test preprocessing: deterministic
    ft_eval_tfm = transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.ToTensor(),
        transforms.Normalize(MEAN, STD),
    ])

    ft_test_ds = ImageDataset(test_paths, np.zeros(len(test_paths), dtype=int), ft_eval_tfm)


# Initialise the globals with the first configured size
setup_finetune(*list(FT_SIZES.items())[0])
print("Fine-tuning plan:", FT_SIZES)
```

```python
def build_ft_model():
    """Create DINOv2 with a new classification head and freeze most of the backbone.

    Only the last ``UNFREEZE_BLOCKS`` transformer blocks, the final norm and the head are trainable.

    Returns:
        The model on ``device``.
    """
    model = timm.create_model(FT_MODEL, pretrained=True, num_classes=N_CLASSES,
                              img_size=FT_IMG_SIZE, drop_path_rate=0.1)

    # Freeze all weights, then unfreeze the last blocks, the final norm and the head
    for p in model.parameters():
        p.requires_grad = False
    for block in model.blocks[-UNFREEZE_BLOCKS:]:
        for p in block.parameters():
            p.requires_grad = True
    for p in model.norm.parameters():
        p.requires_grad = True
    for p in model.head.parameters():
        p.requires_grad = True
    return model.to(device)


def trainable_state(model):
    """Copy only the trainable weights, a small fraction of the full model.

    Args:
        model: Model whose trainable parameters are collected.

    Returns:
        Dict mapping parameter name to a CPU tensor.
    """
    names = {n for n, p in model.named_parameters() if p.requires_grad}
    return {k: v.detach().cpu().clone() for k, v in model.state_dict().items() if k in names}


@torch.no_grad()
def predict_probs(model, loader, tta=True, desc="Predicting"):
    """Predict class probabilities for every image in a loader.

    Args:
        model: Classifier (switched to eval mode).
        loader: DataLoader yielding ``(images, labels)``.
        tta: If True, average predictions over each image and its horizontal mirror.
        desc: Progress-bar label.

    Returns:
        Float64 array of shape ``(n_images, n_classes)``.
    """
    model.eval()
    out = []
    for imgs, _ in tqdm(loader, desc=desc, leave=False):
        imgs = imgs.to(device, non_blocking=True)
        with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
            logits = model(imgs)
            if tta:
                logits = (logits + model(torch.flip(imgs, dims=[3]))) / 2
        out.append(torch.softmax(logits.double(), dim=1).cpu())   # float64 so rows sum to 1 exactly
    return torch.cat(out).numpy()
```

```python
def train_fold(fold_num, tr_idx, va_idx):
    """Fine-tune one fold and save its checkpoint and predictions.

    Trains with early stopping on validation log loss, restores the best epoch and
    writes the best weights, the validation and test probabilities and the epoch
    history. The fold is skipped when its predictions already exist.

    Args:
        fold_num: Fold index (also offsets the random seed).
        tr_idx: Row indices of the training part.
        va_idx: Row indices of the validation part.
    """
    val_file = f"{FT_TAG}_fold{fold_num}_val_probs.npy"
    test_file = f"{FT_TAG}_fold{fold_num}_test_probs.npy"
    ckpt_file = f"{FT_TAG}_fold{fold_num}_best.pt"
    history_file = f"{FT_TAG}_fold{fold_num}_history.json"

    # Resume support: skip folds whose predictions are already saved
    if os.path.exists(val_file) and os.path.exists(test_file):
        print(f"fold {fold_num}: results already saved, skipping")
        return

    set_seed(SEED + fold_num)

    # Data loaders for this fold
    train_ds = ImageDataset([train_paths[i] for i in tr_idx], y[tr_idx], ft_train_tfm)
    val_ds = ImageDataset([train_paths[i] for i in va_idx], y[va_idx], ft_eval_tfm)
    train_loader = DataLoader(train_ds, batch_size=FT_BATCH_SIZE, shuffle=True, drop_last=True,
                              num_workers=NUM_WORKERS, pin_memory=True)
    val_loader = DataLoader(val_ds, batch_size=64, shuffle=False,
                            num_workers=NUM_WORKERS, pin_memory=True)
    test_loader = DataLoader(ft_test_ds, batch_size=64, shuffle=False,
                             num_workers=NUM_WORKERS, pin_memory=True)

    # Model, loss and optimiser
    model = build_ft_model()
    loss_fn = torch.nn.CrossEntropyLoss(label_smoothing=0.1)

    # Discriminative learning rates: higher for the new head, lower for the pretrained blocks
    head_params = list(model.head.parameters())
    head_ids = {id(p) for p in head_params}
    body_params = [p for p in model.parameters() if p.requires_grad and id(p) not in head_ids]
    trainable_params = body_params + head_params
    optimizer = torch.optim.AdamW([
        {"params": body_params, "weight_decay": 0.05},
        {"params": head_params, "weight_decay": 0.01},
    ], lr=BLOCK_LR)
    scheduler = OneCycleLR(optimizer, max_lr=[BLOCK_LR, HEAD_LR],
                           total_steps=FT_EPOCHS * len(train_loader),
                           pct_start=0.15, cycle_momentum=False)
    amp_scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

    # Training loop with early stopping
    best_ll, best_state, bad_epochs = 9e9, None, 0
    history = []

    for epoch in range(1, FT_EPOCHS + 1):
        model.train()
        running_loss, seen = 0.0, 0
        start = time.time()

        bar = tqdm(train_loader, desc=f"fold {fold_num} epoch {epoch}/{FT_EPOCHS}", leave=False)
        for imgs, labels in bar:
            imgs = imgs.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            optimizer.zero_grad(set_to_none=True)
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=use_amp):
                loss = loss_fn(model(imgs), labels)

            amp_scaler.scale(loss).backward()
            amp_scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(trainable_params, 1.0)   # gradient clipping
            amp_scaler.step(optimizer)
            amp_scaler.update()
            scheduler.step()

            running_loss += loss.item() * imgs.size(0)
            seen += imgs.size(0)
            bar.set_postfix(loss=f"{running_loss / seen:.4f}")

        # Evaluate on the held-out fold
        val_probs = predict_probs(model, val_loader, tta=False, desc="validating")
        val_ll = log_loss(y[va_idx], val_probs, labels=list(range(N_CLASSES)))
        val_acc = (val_probs.argmax(axis=1) == y[va_idx]).mean()
        train_loss = running_loss / seen
        print(f"fold {fold_num} | epoch {epoch}/{FT_EPOCHS} | train loss {train_loss:.4f} | "
              f"val log loss {val_ll:.4f} | val acc {val_acc:.3f} | {time.time() - start:.0f}s")
        history.append({"epoch": epoch, "train_loss": train_loss,
                        "val_log_loss": float(val_ll), "val_acc": float(val_acc)})

        # Checkpoint whenever validation log loss improves
        if val_ll < best_ll:
            best_ll, bad_epochs = val_ll, 0
            best_state = trainable_state(model)
            torch.save(best_state, ckpt_file)
            print("   new best, saved", ckpt_file)
        else:
            bad_epochs += 1
            if bad_epochs >= FT_PATIENCE:
                print(f"   no improvement for {FT_PATIENCE} epochs, stopping this fold early")
                break

    # Restore the best epoch; final predictions use flip test-time augmentation
    model.load_state_dict(best_state, strict=False)
    final_val = predict_probs(model, val_loader, tta=True, desc="final val")
    final_test = predict_probs(model, test_loader, tta=True, desc="test")
    np.save(val_file, final_val)
    np.save(test_file, final_test)
    with open(history_file, "w") as f:
        json.dump(history, f, indent=2)

    final_ll = log_loss(y[va_idx], final_val, labels=list(range(N_CLASSES)))
    print(f"fold {fold_num} finished | best-epoch val log loss with mirror averaging: {final_ll:.4f}")

    # Release GPU memory for the next fold
    del model, optimizer, scheduler, train_loader, val_loader, test_loader
    gc.collect()
    torch.cuda.empty_cache()
```

```python
# Fine-tune every fold at every configured size (the longest stage; resumes from saved folds)
if RUN_FINETUNE:
    overall_start = time.time()
    for img_size, epochs in FT_SIZES.items():
        setup_finetune(img_size, epochs)
        print(f"\n===== Fine-tuning at {img_size}px, up to {epochs} epochs per fold =====")
        for fold_num, (tr_idx, va_idx) in enumerate(tqdm(folds, desc=f"Fine-tuning folds @{img_size}")):
            train_fold(fold_num, tr_idx, va_idx)
            print(f"--- {img_size}px: {fold_num + 1}/{N_FOLDS} folds done, {(time.time() - overall_start) / 60:.0f} min so far ---")
else:
    print("Fine-tuning is switched off")
```

```python
# Collect fold results: stitch validation predictions into OOF predictions, average the
# fold models' test predictions, and register each size as a calibrated blend component
if RUN_FINETUNE:
    for img_size, epochs in FT_SIZES.items():
        setup_finetune(img_size, epochs)
        oof_ft = np.zeros((len(y), N_CLASSES))
        test_ft = np.zeros((len(test_paths), N_CLASSES))
        complete = True

        for fold_num, (tr_idx, va_idx) in enumerate(folds):
            val_file = f"{FT_TAG}_fold{fold_num}_val_probs.npy"
            test_file = f"{FT_TAG}_fold{fold_num}_test_probs.npy"
            if not (os.path.exists(val_file) and os.path.exists(test_file)):
                print(f"{img_size}px: fold {fold_num} has no saved results, leaving this model out")
                complete = False
                break
            val_p = np.load(val_file).astype(np.float64)
            oof_ft[va_idx] = val_p / val_p.sum(axis=1, keepdims=True)
            test_p = np.load(test_file).astype(np.float64)
            test_ft += (test_p / test_p.sum(axis=1, keepdims=True)) / N_FOLDS

        if complete:
            print(f"Fine-tuned @{img_size}px OOF log loss (raw):", round(score(oof_ft), 4))
            add_component(f"finetuned_{img_size}", oof_ft, test_ft)
```

## Step 9: Blending
Every model is now calibrated. The blend is a weighted average of their probabilities, with weights searched on the out-of-fold predictions. Only a handful of numbers are fitted, which leaves little room to overfit, and giving one model all the weight is one of the candidates, so the blend cannot score worse OOF than the best single model. The blend is then calibrated once more.

```python
# Grid search for blend weights on the out-of-fold predictions
names = list(components)
print("Blending:", names)

# Weight grid: steps of 0.1 that sum to 1
grid = [w for w in itertools.product(range(11), repeat=len(names)) if sum(w) == 10]
print("Weight combinations to try:", len(grid))

best_w, best_blend_ll = None, 9e9
for w in tqdm(grid, desc="Blend weights"):
    blended = sum((wi / 10) * components[n][0] for wi, n in zip(w, names))
    ll = score(blended)
    if ll < best_blend_ll:
        best_w, best_blend_ll = w, ll

print("Best weights:", {n: wi / 10 for n, wi in zip(names, best_w)})
print("Blend OOF log loss:", round(best_blend_ll, 4))

blend_oof = sum((wi / 10) * components[n][0] for wi, n in zip(best_w, names))
blend_test = sum((wi / 10) * components[n][1] for wi, n in zip(best_w, names))
```

```python
# Final calibration of the blend
temp, eps, final_ll = fit_calibration(blend_oof, label="Final calibration")
final_oof = calibrate(blend_oof, temp, eps)
final_test = calibrate(blend_test, temp, eps)

# Blend versus each calibrated component
for n in components:
    print(f"{n:12s} OOF log loss {score(components[n][0]):.4f}")
best_single_ll = min(score(components[n][0]) for n in components)
print(f"{'blend':12s} OOF log loss {final_ll:.4f}  (temp={temp:.2f}, eps={eps})")
print(f"Improvement over the best single model: {best_single_ll - final_ll:.4f}")

# Per-fold log loss (large gaps indicate harder camera sites)
for i, (tr_idx, va_idx) in enumerate(folds):
    fold_ll = log_loss(y[va_idx], final_oof[va_idx], labels=list(range(N_CLASSES)))
    print(f"fold {i} log loss: {fold_ll:.4f}")

np.save("part3_final_oof.npy", final_oof)
np.save("part3_final_test.npy", final_test)
print("Saved part3_final_oof.npy and part3_final_test.npy")
```
